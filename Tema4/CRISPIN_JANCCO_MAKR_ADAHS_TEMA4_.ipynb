```python
import numpy as np
import pandas as pd

datos = {
    "nombre": ["Ana", "Carlos", "Beatriz", "Diego", "Elena", "Fernando"],
    "curso": ["Programación", "Estadística", "Programación", "Estadística", "Programación", "Estadística"],
    "nota": [15.5, 18.0, np.nan, 12.0, 16.5, np.nan],  
    "asistencia_pct": [85.0, np.nan, 90.0, 75.0, 95.0, 80.0] 
}

df_original = pd.DataFrame(datos)
print("df_original",df_original)

```

```python
df = pd.read_csv("estudiantes.csv")

print("Forma del DataFrame (filas, columnas):", df.shape)
print("\nCantidad de valores faltantes por columna:")
print(df.isnull().sum())
```

```python
df_limpio = df.dropna(subset=["nota"]).copy()

promedio_asistencia = df_limpio["asistencia_pct"].mean()
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(promedio_asistencia)

print("Valores faltantes tras la limpieza:")
print(df_limpio.isnull().sum())
```

```python
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()
print("\nPromedio de nota agrupado por curso:")
print(promedio_por_curso)
```

```python
df_limpio.to_csv("estudiantes_limpio.csv", index=False)
```
