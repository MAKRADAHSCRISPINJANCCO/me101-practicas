```python
import numpy as np
import matplotlib.pyplot as plt


fila = np.linspace(0, 255, 5, dtype=np.uint8)
gris = np.tile(fila, (5, 1))

print("Imagen:")
print(gris)
print("Shape:", gris.shape)
print("Tipo de dato:", gris.dtype)
print("Media:", gris.mean())
```

```python
plt.figure(figsize=(6, 6))                   
plt.imshow(gris, cmap="gray", vmin=0, vmax=255,
           interpolation="nearest")           
plt.title("Degradado horizontal 5×5", fontsize=14)
plt.colorbar(label="Intensidad (0-255)")
plt.xticks(range(5)); plt.yticks(range(5))    
plt.grid(color="red", linewidth=0.5, alpha=0.4) 
plt.savefig("gris_tema6.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
recorte = gris[:, :] 
print("Recorte shape:", recorte.shape)
print(recorte)
```

```python
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [255, 0, 0]
rgb[:, 2:4] = [0, 0, 255]
print("Dimensiones:", rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])
plt.imshow(rgb)
plt.title("Imagen RGB: rojo y azul")
plt.axis("off")
plt.show()
```

```python
R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

print("Canal rojo:")
print(R)

print("Canal verde:")
print(G)

print("Canal azul:")
print(B)
```

```python
gris_promedio = np.mean(rgb, axis=2)
gris_luminosidad = (
    0.299 * rgb[:, :, 0]
    + 0.587 * rgb[:, :, 1]
    + 0.114 * rgb[:, :, 2]
)

print("Gris por promedio:")
print(gris_promedio)
print("Gris por luminosidad:")
print(gris_luminosidad)
```

```python
umbral = 50
binaria = np.where(gris_luminosidad > umbral, 255, 0).astype(np.uint8)

print("Imagen binarizada:")
print(binaria)

plt.figure(figsize=(6, 6))
plt.imshow(binaria, cmap="gray", vmin=0, vmax=255,
           interpolation="nearest")
plt.title(f"Binarizada (umbral = {umbral})", fontsize=14)
plt.xticks(range(4)); plt.yticks(range(4))
plt.grid(color="red", linewidth=1, alpha=0.6)
plt.show()
```
