```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    'PM10': [45.0, 28.0, 62.0, 35.0, np.nan],
    'NO2':  [18.2, 12.5, 25.4, 15.0, 9.8],
    'Ruido': [72, 58, 81, 65, 52]
}, index=['Centro', 'San Jerónimo', 'Wanchaq', 'San Sebastián', 'Poroy'])

df_clean = df.fillna(df.median())   # PM10 de Poroy = 40.0
corr = df_clean.corr()
print(corr.round(4))
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True)
plt.title("Matriz de correlación (seaborn)")
plt.tight_layout()
plt.savefig("heatmap_seaborn.png", dpi=150)
plt.show()
```

```python
mascara = np.triu(np.ones_like(corr, dtype=bool))
print("Celdas visibles:", int((~mascara).sum()))   # 3

plt.figure(figsize=(6, 5))
sns.heatmap(corr, mask=mascara, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True)
plt.title("Matriz de correlación con máscara (seaborn)")
plt.tight_layout()
plt.savefig("heatmap_seaborn_mascara.png", dpi=150)
plt.show()
```
