```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")


media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()              
var_poblacional = np.var(df["nota"])         
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print("Media:", round(media, 4))
print("Mediana:", round(mediana, 4))
print("Desv. std muestral:", round(std_muestral, 4))
print("Var. poblacional:", round(var_poblacional, 4))
print("Q1:", round(q1, 4))
print("Q3:", round(q3, 4))
print("IQR:", round(iqr, 4))
```

```python
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()
skew_scipy = stats.skew(df["nota"])
kurt_scipy = stats.kurtosis(df["nota"])

print("Skewness pandas:", round(skew_pandas, 4))
print("Skewness scipy:", round(skew_scipy, 4))
print("Kurtosis pandas:", round(kurt_pandas, 4))
print("Kurtosis scipy:", round(kurt_scipy, 4))
```

```python
def resumen_estadistico(serie, decimales=4):
    n = int(serie.count())
    media = serie.mean()
    mediana = serie.median()
    desv_std = serie.std()
    cv_pct = desv_std / media * 100
    return {
        "n": n,
        "media": round(float(media), decimales),
        "mediana": round(float(mediana), decimales),
        "desv_std": round(float(desv_std), decimales),
        "cv_pct": round(float(cv_pct), decimales),
    }

print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(resumen["cv_pct"])
    print(columna, resumen, "-> Dispersión:", nivel)
```
